```python
#W5A3
def primeNum(a):
  for i in range(2, int(a**0.5)+1):
    if a%i == 0:
      return False
    return True

a = int(input())
if primeNum(a):
  print("YES")
else:
  print("NO")

```

```python
#W5A4
def perfectNum(a):
    if a <= 1:
        print("No")
        return

    s = 1
    for i in range(2, int(a**0.5) + 1):
        if a % i == 0:
            s += i
            if i != a // i:
                s += a // i

    if s == a:
        print("Yes")
    else:
        print("No")

n = int(input())
perfectNum(n)
```

```python
#W5A5
def findNum(a, x):
  for i in range(0, len(a)):
    if x == a[i]:
      return i+1
  return -1
a = list(map(int, input().split()))
k = int(input('Nhap so k: '))
print(findNum(a, k))
```

```python
#W5A6
def giaithua(n):
  ans = 1
  for i in range(1,n+1):
    ans = ans*i
  return ans
n = int(input())
print(f'n giai thua la {giaithua(n)}')
```

```python
#W5A9
def TongCacChuSo(n):
  sum = 0
  for i in range (len(n)):
    sum = sum + int(n[i])
  return sum

n = input()
print(TongCacChuSo(n))
```

```python
#W5A10
def Dang_cau(a,b):
    map_ab = {}
    mapped_b = set()
    for i in range(len(a)):
        c1 = a[i]
        c2 = b[i]
        d = ord(c2)
        if c1 in map_ab:
            if map_ab[c1] != d:
                return False
        else:
            if d in mapped_b:
                return False
            map_ab[c1] = d 
            mapped_b.add(d)
    return True
    
a = input()
b = input()
if Dang_cau(a, b):
    print('Dang cau')
else:
    print('Khong dang cau')
```
