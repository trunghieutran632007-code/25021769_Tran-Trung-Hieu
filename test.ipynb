```python
def Phan_tich(n):
    num2 = 2
    a = []
    while n!= 1:
        if n%num2 == 0:
            n = n/num2
            a.append(num2)
        else:
            num2 +=1
    return a

print(Phan_tich(120))
a = [str(x) for x in Phan_tich(120) if x%2 == 0]
s = ','.join(a)
print(s)
    
        
```

```python
def Di_chuyen(s):
    Toa_do = [0,0]
    path = []
    for c in s:
        if c == "R":
            Toa_do[0] += 1
            path.append(Toa_do[:])
        if c == "L":
            Toa_do[0] -= 1
            path.append(Toa_do[:])
        if c == "U":
            Toa_do[1] += 1
            path.append(Toa_do[:])
        if c == "D":
            Toa_do[1] -= 1
            path.append(Toa_do[:])
    return path
def Di_qua(path):
    visited = {(0,0)}
    path_tup = [tuple(x) for x in path]
    for point in path_tup:
        if point in visited:
            return True
        else:
            visited.add(point)
    return False

s = "UURRDL"
print(Di_chuyen(s))
print(Di_qua(Di_chuyen(s)))

```
