```python
def Di_chuyen(s):
    Toa_do = [0,0]
    path = []
    for c in s:
        if c == "R":
            Toa_do[0] += 1
            path.append(Toa_do[:])
        if c == "L":
            Toa_do[0] -= 1
            path.append(Toa_do[:])
        if c == "U":
            Toa_do[1] += 1
            path.append(Toa_do[:])
        if c == "D":
            Toa_do[1] -= 1
            path.append(Toa_do[:])
    return path
def Di_qua(path):
    visited = {(0,0)}
    path_tup = [tuple(x) for x in path]
    for point in path_tup:
        if point in visited:
            return True
        else:
            visited.add(point)
    return False

s = "UURRDL"
print(Di_chuyen(s))
print(Di_qua(Di_chuyen(s)))

```

```python
#Bai tap ve khoang cach
s = input()
vi_tri = {}
for i in range(len(s)):
    char = s[i]
    if char not in vi_tri:
        vi_tri[char] = [i]
    else:
        vi_tri[char].append(i)
max_dist = 0
res_char = None
sorted_char = sorted(vi_tri.keys())
for char in sorted_char:
    indices = vi_tri[char]

    if len(indices) < 2:
        continue
    cur_max_dist = 0
    for i in range(len(indices)-1):
        dist = indices[i+1] - indices[i]
        if dist > cur_max_dist:
            cur_max_dist = dist
            
    
    if cur_max_dist > max_dist:
        max_dist = cur_max_dist
        res_char = char
if res_char is None:
    print(0)
else:
    print(f"{res_char} {max_dist}")
```

```python
#Bai tap so phuc
import math
class ComplexNumber:
    def __init__(self, initRealPart, initImaginaryPart):
        self.real = float(initRealPart)
        self.image = float(initImaginaryPart)
    def __add__(self, other):
        return ComplexNumber(self.real + other.real, self.image + other.image)
    def __mul__(self, other):
        new_real = self.real * other.real - self.image * other.image
        new_imag = self.real * other.image + self.image * other.real
        return ComplexNumber(new_real, new_imag)
    def __abs__(self):
        return math.sqrt(self.real**2 + self.image**2)
    def conjugate(self):
        return ComplexNumber(self.real, -self.image)
    def __str__(self):
        if self.image >= 0:
            return (f"{self.real} + {self.image}i")
        else:
            return (f"{self.real} {self.image}i")

if __name__ == "__main__":
    line1 = input().strip().split()
    z1 = ComplexNumber(float(line1[0]), float(line1[1]))
    print(z1)

    line2 = input().strip().split()
    z2 = ComplexNumber(float(line2[0]), float(line2[1]))
    print(z2)

    tong = z1 + z2
    print(tong)

    tich = z1*z2
    print(tich)

    modun = abs(tich)
    print(f"{modun:.2f}")

    lienhop = tich.conjugate()
    print(lienhop)



    
    
        
```
