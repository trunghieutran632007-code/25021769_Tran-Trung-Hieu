```python
#W11A5
from pathlib import Path
WORKDIR = Path(".")

def count(path: str, character: str):
    count = 0
    file = WORKDIR/path
    with open(file, "r") as f:
        s = f.read()
        for c in s:
            if c == character:
                count += 1
    return count

#main
s = input("Nhap chuoi: ")
char = input("Nhap ki tu: ")
#Tao file va luu chuoi s vao file (Bai nay e tao file bang 'w', nhung bai khac e tao file tren folder)
with open(WORKDIR/"th13_5.txt", "w") as f:
    f.write(s)

    
print(count("th13_5.txt", char))

```

```python
#W11A6
from pathlib import Path
WORKDIR = Path(".")
def countCharacter(path: str, letter: str):
    count = 0
    file = WORKDIR/path
    with open(file, "r") as f:
        s = f.read()
        s = s.lower()
        for c in s:
            if c == letter:
                count += 1
    return count
#main
letter = "e"
print(countCharacter("th13_6.txt", letter))
```

```python
#W11A8
def fibonacci_words(n):
    dp = [""] * (n + 1)

    dp[0] = "a"
    if n >= 1:
        dp[1] = "b"

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp


def write_to_file(n, filename):
    fib_words = fibonacci_words(n)

    with open(filename, "w") as f:
        for i in range(n + 1):
            f.write(f"f({i}) = {fib_words[i]}\n")
            print(fib_words[i]) #In ra man hinh de theo doi
#main
write_to_file(20, "th13_8.txt")

```

```python
#W11A9
def grade10(path: str):
    try:
        with open(path, "r") as f:
            
            lines = f.readlines()
            for i in range(1, len(lines)):
                line = lines[i].strip() 

                parts = line.split(',')
                if len(parts) > 5:
                    thcs4 = parts[5].strip() 
                    
                    if thcs4 == '10':
                        ho = parts[1].strip()  
                        ten = parts[2].strip() 
                        print(f"{ho} {ten}")

    except FileNotFoundError:
        print(path)
    except Exception as e:
        print(e)

#main
content = """STT,Ho,Ten,Ngay sinh,Dai so,THCS4,THCS1
1,Truong Vo,Ky,10/10/1999,8,10,9
2,Chu Chi,Nhuoc,13/9/1999,10,10,10
3,Nguyen Van,A,01/01/2000,9,8,7
4,Le Thi,B,02/02/2000,7,10,8
"""

filename = "danh_sach_sv.csv"
with open(filename, "w", ) as f:
    f.write(content)
print("Danh sách sinh viên có điểm THCS4 là 10:")
grade10(filename)
```
