```python
#W11A6
from pathlib import Path
WORKDIR = Path(".")
def countCharacter(path: str, letter: str):
    count = 0
    file = WORKDIR/path
    with open(file, "r") as f:
        s = f.read()
        s = s.lower()
        for c in s:
            if c == letter:
                count += 1
    return count
#main
letter = "e"
print(countCharacter("th13_6.txt", letter))
```

```python
#W11A8
def fibonacci_words(n):
    dp = [""] * (n + 1)

    dp[0] = "a"
    if n >= 1:
        dp[1] = "b"

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp


def write_to_file(n, filename):
    fib_words = fibonacci_words(n)

    with open(filename, "w") as f:
        for i in range(n + 1):
            f.write(f"f({i}) = {fib_words[i]}\n")
            print(fib_words[i]) #In ra man hinh de theo doi
#main
write_to_file(20, "th13_8.txt")

```

```python
#W11A9
def grade10(path: str):
    try:
        with open(path, "r") as f:
            
            lines = f.readlines()
            for i in range(1, len(lines)):
                line = lines[i].strip() 

                parts = line.split(',')
                if len(parts) > 5:
                    thcs4 = parts[5].strip() 
                    
                    if thcs4 == '10':
                        ho = parts[1].strip()  
                        ten = parts[2].strip() 
                        print(f"{ho} {ten}")

    except FileNotFoundError:
        print(path)
    except Exception as e:
        print(e)

#main
content = """STT,Ho,Ten,Ngay sinh,Dai so,THCS4,THCS1
1,Truong Vo,Ky,10/10/1999,8,10,9
2,Chu Chi,Nhuoc,13/9/1999,10,10,10
3,Nguyen Van,A,01/01/2000,9,8,7
4,Le Thi,B,02/02/2000,7,10,8
"""

filename = "danh_sach_sv.csv"
with open(filename, "w", ) as f:
    f.write(content)
print("Danh sách sinh viên có điểm THCS4 là 10:")
grade10(filename)
```

```python
#W11A12
from pathlib import Path
WORKDIR = Path(".")
def maximumProduct (path: str) -> int:
    with open(path, "r") as f:
        n = f.readline()
        nums = f.readline()
    nums = nums.split()
    if len(nums) < 3: return -1
    nums = [int(x) for x in nums]
    nums.sort()
    option1 = nums[-1] * nums[-2] * nums[-3]
    option2 = nums[0] * nums[1] * nums[-1]
    return max(option1, option2)
#main
filename = WORKDIR/"th13_12.txt"
maximumProduct(filename)
```

```python
#11A13
def moveZeroes(fileName: str) -> list:
    try:
        with open(fileName, 'r') as f:
            line1 = f.readline()
            if not line1: return []
            n = int(line1.strip())
            
            line2 = f.readline()
            if not line2: return []
            arr = list(map(int, line2.split()))


        pos = 0
        
        for i in range(len(arr)):
            if arr[i] != 0:
                arr[pos], arr[i] = arr[i], arr[pos]
                pos += 1
        
        return arr

    except Exception:
        print("Error")
        

#Tao file
if __name__ == "__main__":
    fn = "th13_13.txt"
    with open(fn, "w") as f:
        f.write("6\n")             
        f.write("0 1 0 3 12 0") 
    
    # Chạy thử hàm
    result = moveZeroes(fn)
    print(result) 
    """
    0 1 0 3 12 0
    1 0 0 3 12 0
    1 3 0 0 12 0
    1 3 12 0 0 0
    """
```

```python
#W11A14
def averageTime(path: str):
    total_time = 0
    count = 0
    
    try:
        with open(path, 'r') as file:
            # Đọc dòng đầu tiên
            header = file.readline()
        
            for line in file:
                line = line.strip()
        
                if line:
                    parts = line.split(',')
                    
                    time_value = float(parts[1])
                    
                    total_time += time_value
                    count += 1
        
        
        if count == 0:
            return 0.0
            
        average = total_time / count
        
        return round(average, 2)

    except FileNotFoundError:
        return "Không tìm thấy file"
    except Exception as e:
        return f"{e}"



# Tạo file th13_14.csv
test_filename = "th13_14.csv"
with open(test_filename, "w") as f:
    f.write("Lap,Time\n")
    f.write("1,143.23\n")
    f.write("2,134.12\n")
    f.write("3,140.50\n")

# Gọi hàm và in kết quả
ket_qua = averageTime(test_filename)
print(f"AVR: {ket_qua}")
```

```python
#W11A15
"""
Thành phố X mở một rạp chiếu phim mới, rất nhiều người muốn đến xem thử và hưởng các khuyến mại trong ngày khai trương.

Để đưa cho khách hàng các lựa chọn, rạp chiếu phim treo các poster về phim cùng với điểm đánh giá IMDB (rating) và mô tả của khách hàng về bộ phim đó.

Hãy viết hàm def findMovies (path: str) -> list[str] giúp khán giả tìm ra phim với điểm đánh giá lớn hơn 8.0 và có mô tả Không phải là "nhàm chán" ("boring").

Hàm nhận đầu vào là đường dẫn đến tệp chứa các poster phim. Hàm trả lại các tên những phim thỏa mãn điều kiện trên. Nếu không tìm thấy phim nào phù hợp, trả về thông báo "All are boring!". Kết quả in theo thứ tự xuất hiện trước sau trong file đầu vào

Tệp có cấu trúc như sau:

id, movie, description, rating
1, Thor, great 3D, 8.0
2, The Rings, Fantasy, 7.5
3, Irish, boring, 6.2
4, Ice song, Fantasy, 8.6
5, Home Alone, Interesting, 6.5
"""
from pathlib import Path
WORKDIR = Path(".")

def findMovies (path: str) -> list[str]:
    res = []
    with open(WORKDIR/path) as f:
        head = f.readline()
        for line in f:
            if line:
                line = line.split(",")
                curr_rate = float(line[3].strip())
                curr_des = line[2].strip()
                if curr_rate > 8 and curr_des != "boring":
                    res.append(line[1].strip())
    if not res: return "All are boring!"
    return res

#Tạo file th13_15.csv
content = """
id, movie, description, rating
1, Thor, great 3D, 8.0
2, The Rings, Fantasy, 7.5
3, Irish, boring, 6.2
4, Ice song, Fantasy, 8.6
5, Home Alone, Interesting, 6.5
""".strip()
with open(WORKDIR/"th13_15.csv", "w") as f:
    f.write(content)

#main
print(findMovies("th13_15.csv"))



```
