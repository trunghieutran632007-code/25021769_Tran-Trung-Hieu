```python
#W11A8
def fibonacci_words(n):
    dp = [""] * (n + 1)

    dp[0] = "a"
    if n >= 1:
        dp[1] = "b"

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp


def write_to_file(n, filename):
    fib_words = fibonacci_words(n)

    with open(filename, "w") as f:
        for i in range(n + 1):
            f.write(f"f({i}) = {fib_words[i]}\n")
            print(fib_words[i]) #In ra man hinh de theo doi
#main
write_to_file(20, "th13_8.txt")

```

```python
#W11A12
from pathlib import Path
WORKDIR = Path(".")
def maximumProduct (path: str) -> int:
    with open(path, "r") as f:
        n = f.readline()
        nums = f.readline()
    nums = nums.split()
    if len(nums) < 3: return -1
    nums = [int(x) for x in nums]
    nums.sort()
    option1 = nums[-1] * nums[-2] * nums[-3]
    option2 = nums[0] * nums[1] * nums[-1]
    return max(option1, option2)
#main
filename = WORKDIR/"th13_12.txt"
maximumProduct(filename)
```

```python
#11A13
def moveZeroes(fileName: str) -> list:
    try:
        with open(fileName, 'r') as f:
            line1 = f.readline()
            if not line1: return []
            n = int(line1.strip())
            
            line2 = f.readline()
            if not line2: return []
            arr = list(map(int, line2.split()))


        pos = 0
        
        for i in range(len(arr)):
            if arr[i] != 0:
                arr[pos], arr[i] = arr[i], arr[pos]
                pos += 1
        
        return arr

    except Exception:
        print("Error")
        

#Tao file
if __name__ == "__main__":
    fn = "th13_13.txt"
    with open(fn, "w") as f:
        f.write("6\n")             
        f.write("0 1 0 3 12 0") 
    
    # Chạy thử hàm
    result = moveZeroes(fn)
    print(result) 
    """
    0 1 0 3 12 0
    1 0 0 3 12 0
    1 3 0 0 12 0
    1 3 12 0 0 0
    """
```

```python
#W11A14
def averageTime(path: str):
    total_time = 0
    count = 0
    
    try:
        with open(path, 'r') as file:
            # Đọc dòng đầu tiên
            header = file.readline()
        
            for line in file:
                line = line.strip()
        
                if line:
                    parts = line.split(',')
                    
                    time_value = float(parts[1])
                    
                    total_time += time_value
                    count += 1
        
        
        if count == 0:
            return 0.0
            
        average = total_time / count
        
        return round(average, 2)

    except FileNotFoundError:
        return "Không tìm thấy file"
    except Exception as e:
        return f"{e}"



# Tạo file th13_14.csv
test_filename = "th13_14.csv"
with open(test_filename, "w") as f:
    f.write("Lap,Time\n")
    f.write("1,143.23\n")
    f.write("2,134.12\n")
    f.write("3,140.50\n")

# Gọi hàm và in kết quả
ket_qua = averageTime(test_filename)
print(f"AVR: {ket_qua}")
```

```python
#W11A16
import math
from pathlib import Path
WORKDIR = Path(".")

def tinh_chu_vi_tu_file(duong_dan_file):
    try:

        with open(duong_dan_file, 'r') as f:
            content = f.read().strip()

        #content = content.replace('\r\n', '\n')
        cac_khoi_hinh = content.split('\n\n')

        for khoi in cac_khoi_hinh:
            lines = khoi.strip().split('\n')
            
            if len(lines) < 2:
                continue


            loai_hinh = lines[0].split(':')[1].strip()
            
            thong_so_line = lines[1].strip()
            
            chu_vi = 0.0

            if loai_hinh == 'SQUARE':
                # Format: "SIDE: 5.3"
                canh = float(thong_so_line.split(':')[1])
                chu_vi = 4 * canh

            elif loai_hinh == 'RECTANGLE':
                # Format: "WIDTH: 5.0 HEIGHT: 16.0"
                # ['WIDTH:', '5.0', 'HEIGHT:', '16.0']
                parts = thong_so_line.split()
                rong = float(parts[1])
                cao = float(parts[3])
                chu_vi = 2 * (rong + cao)

            elif loai_hinh == 'CIRCLE':
                # Format: "RADIUS: 2.15"
                ban_kinh = float(thong_so_line.split(':')[1])
                chu_vi = 2 * math.pi * ban_kinh

            print(f"{round(chu_vi, 2)}")

    except Exception as e:
        print(f"{e}")

# Tao file
if __name__ == "__main__":
    ten_file_mau = "th13_16.txt"
    noi_dung_mau = """TYPE: SQUARE
SIDE: 5.3

TYPE: RECTANGLE
WIDTH: 5.0 HEIGHT: 16.0

TYPE: CIRCLE
RADIUS: 2.15""".strip()
    
    
    with open(WORKDIR/ten_file_mau, "w") as f:
        f.write(noi_dung_mau)
    
    
    
    # Yêu cầu nhập đường dẫn (như đề bài)
    path = input("Nhap duong dan: ")
    tinh_chu_vi_tu_file(path)
```

```python
#W11A17
import os

def productExceptSelf(path: str) -> list[int]:
    """
    Hàm đọc dãy số từ file và trả về dãy số mới
    trong đó mỗi phần tử là tích của tất cả các phần tử khác trừ chính nó.
    """
    try:
        with open(path, 'r') as f:
            content = f.read()
            nums = [int(x) for x in content.split()]
    except Exception as e:
        print(e)

    n = len(nums)
    
    if n <= 1:
        return []

    
    result = [1] * n

    left_product = 1
    for i in range(n):
        result[i] = left_product
        left_product *= nums[i]
        #print(result)

    right_product = 1
    for i in range(n - 1, -1, -1):
        result[i] *= right_product
        right_product *= nums[i]

    print(nums)
    print(result)
    
    return result

#main
filename = "th13_17.txt"
with open(filename, "w") as f:
    f.write("2 4 6 8")


output = productExceptSelf(filename)

```

```python
#W11A18
import math

def estimatedTime(path: str):
    try:
        
        with open(path, 'r', encoding='utf-8') as f:
            lines = f.readlines()

        
        if len(lines) < 3:
            return "File không đúng định dạng mẫu"

        
        line_src = lines[0].strip().split()
        src_x = float(line_src[2])
        src_y = float(line_src[4])

        
        line_des = lines[1].strip().split()
        des_x = float(line_des[2])
        des_y = float(line_des[4])

        
        line_vel = lines[2].strip().split()
        velocity = float(line_vel[1])

     
        distance = math.sqrt((des_x - src_x)**2 + (des_y - src_y)**2)

        time_needed = distance / velocity

        return round(time_needed, 2)

    except FileNotFoundError:
        return "Không tìm thấy file"
    except Exception as e:
        return f"Có lỗi xảy ra: {e}"

#Tao file
content = """SRC X: 0 Y: 0
DES X: 10000 Y: 10000
VELOCITY: 6 KM/S""".strip()

filename = "th13_18.txt"
with open(filename, "w") as f:
    f.write(content)


result = estimatedTime(filename)
print(result)
```

```python
#W11A19 Tham khảo AI
import struct
import os

def read_binary_doubles():
    # 1. Nhập đường dẫn tệp từ bàn phím
    # Sử dụng .strip() để loại bỏ khoảng trắng thừa nếu có
    file_path = input("Nhập đường dẫn đến tệp nhị phân: ").strip()

    # Kiểm tra xem file có tồn tại không
    if not os.path.exists(file_path):
        print(f"Lỗi: Không tìm thấy tệp tại đường dẫn '{file_path}'")
        return

    try:
        # 2. Mở file ở chế độ đọc nhị phân ('rb' - read binary)
        with open(file_path, 'rb') as f:
            while True:
                # 3. Đọc 8 bytes (kích thước chuẩn của một số kiểu double là 64-bit = 8 bytes)
                byte_data = f.read(8)

                # Nếu dữ liệu đọc được nhỏ hơn 8 bytes, nghĩa là đã hết file
                if len(byte_data) < 8:
                    break

                # 4. Giải mã (unpack) chuỗi bytes thành số thực
                # 'd' đại diện cho kiểu double trong module struct
                # hàm unpack trả về một tuple, nên ta lấy phần tử đầu tiên [0]
                value = struct.unpack('d', byte_data)[0]

                # 5. In ra màn hình, làm tròn 5 chữ số thập phân
                # :.5f là định dạng float với 5 số sau dấu phẩy
                print(f"{value:.5f}")

    except Exception as e:
        print(f"Đã xảy ra lỗi khi đọc file: {e}")

if __name__ == "__main__":
    read_binary_doubles()
```

```python
import struct

# Dữ liệu giả lập
numbers = [1.23456789, 10.0, -5.555555, 3.1415926535, 100.000001]

# Tạo file nhị phân
filename = 'data.bin'
with open(filename, 'wb') as f:
    for num in numbers:
        # Đóng gói số thực thành 8 bytes nhị phân và ghi vào file
        f.write(struct.pack('d', num))

print(f"Đã tạo file '{filename}' thành công. Hãy dùng đường dẫn này để test bài trên.")
```

```python
#W11A20
def imageSmoother(path: str):
    try:
        with open(path, 'r') as f:
            lines = f.read().strip().split('\n')
            
        if not lines:
            print("File rỗng")
            return
        M = int(lines[0].strip())
        

        img = []
        for i in range(1, M + 1):
            row = list(map(int, lines[i].strip().split()))
            img.append(row)

        res = [[0] * M for _ in range(M)]

        for r in range(M):
            for c in range(M):
                total = 0
                count = 0
                for i in range(r - 1, r + 2):
                    for j in range(c - 1, c + 2):
                        if 0 <= i < M and 0 <= j < M:
                            total += img[i][j]
                            count += 1
                
                res[r][c] = total // count


        print("Ma trận sau khi làm mịn:")
        for row in res:
            print(*(row)) #Bỏ dấu ngoặc vuông và dấu , khi in

    except FileNotFoundError:
        print(f"Không tìm thấy file tại đường dẫn: {path}")
    except Exception as e:
        print(f"Có lỗi xảy ra: {e}")

if __name__ == "__main__":
    # Tạo file
    filename = "th13_20.txt"
    content = """3
1 2 3
4 5 6
1 1 1""".strip()
    
    with open(filename, "w") as f:
        f.write(content)

    # Gọi hàm
    imageSmoother(filename)
```
