```python
#W6A16
n = int(input("Nhap hang: "))
m = int(input("Nhap cot: "))
matrix = []
for i in range(n):
    row = []
    for j in range(m):
       y = int(input(f"Nhap so thu {j+1} cua hang thu {i+1}"))
       row.append(y)
    matrix.append(row)
print(matrix) 

for row in matrix:
    for item in row:
            # {:>4} nghĩa là: dành 4 chỗ trống, căn lề phải (>)
        print(f"{item:>4}", end="") 
    print()
```
