```python
#W6A18
m, n, k = map(int, input().split())
matrix = []
#Nhap matrix
for i in range(n):
    row = []
    for j in range(m):
       y = int(input(f"Nhap so thu {j+1} cua hang thu {i+1}"))
       row.append(y)
    matrix.append(row)
print(matrix)

for row in matrix:
    for item in row:
            # {:>4} nghĩa là: dành 4 chỗ trống, căn lề phải (>)
        print(f"{item:>4}", end="") 
    print()
sumcol = 0
sumrow = 0
for i in matrix[k-1]:
    sumrow += i
print(f'Tong hang {k} la: {sumrow}')

for i in matrix:
    sumcol += i[k-1]
print(f'Tong cot {k} la: {sumcol}')

```

```python
#W6A20
def vi_tri(a, k):
    for i in range(len(a)):
        if a[i] == k:
            return i
    return -1
a = list(map(int, input().split()))
k = int(input())
print(f'vi tri do la: a[{vi_tri(a, k)}]')
```

```python
#W6A23
def Delete_num(a, x):
    # a[:] tạo ra một bản sao (copy) của list a để duyệt
    for i in a[:]: 
        if i < x:
            a.remove(i)
    return a

a = list(map(int, input().split()))
print(a)
x = int(input("Nhap so x: "))
print(x)
print(Delete_num(a, x))
```
