```python
#W6A16
n = int(input("Nhap hang: "))
m = int(input("Nhap cot: "))
matrix = []
for i in range(n):
    row = []
    for j in range(m):
       y = int(input(f"Nhap so thu {j+1} cua hang thu {i+1}"))
       row.append(y)
    matrix.append(row)
print(matrix) 

for row in matrix:
    for item in row:
            # {:>4} nghĩa là: dành 4 chỗ trống, căn lề phải (>)
        print(f"{item:>4}", end="") 
    print()
```

```python
#W6A17
n = int(input("Nhap matrix vuong co n: "))
matrix = []
for i in range(n):
    row = []
    for j in range(n):
       y = int(input(f"Nhap so thu {j+1} cua hang thu {i+1}"))
       row.append(y)
    matrix.append(row)
print(matrix) 
for i in range(n):
    print(matrix[i][i])

```

```python
#W6A20
def vi_tri(a, k):
    for i in range(len(a)):
        if a[i] == k:
            return i
    return -1
a = list(map(int, input().split()))
k = int(input())
print(f'vi tri do la: a[{vi_tri(a, k)}]')
```

```python
#W6A21
a = list(map(str, input().split()))
dic = {}
for i in range(len(a)):
    if a[i] not in dic:
        dic[a[i]] = []
        dic[a[i]].append(i)
    else:
        dic[a[i]].append(i)
for key, value in dic.items():
    dic[key] = tuple(dic[key])

print(dic)
```
