```python
#W11A3
from pathlib import Path
WORKDIR = Path(".")
res = []
def housesOfHogwarts(path: str):
    file = WORKDIR/path
    with open(file, "r") as f:
        num = int(f.readline().strip())
        for _ in range(num):
            line = f.readline()
            res.append(line)
    return res
#print(housesOfHogwarts("th13_3.txt"))
lstOfName = housesOfHogwarts("th13_3.txt")
for name in lstOfName:
    print(name)


```

```python
#W11A5
from pathlib import Path
WORKDIR = Path(".")

def count(path: str, character: str):
    count = 0
    file = WORKDIR/path
    with open(file, "r") as f:
        s = f.read()
        for c in s:
            if c == character:
                count += 1
    return count

#main
s = input("Nhap chuoi: ")
char = input("Nhap ki tu: ")
#Tao file va luu chuoi s vao file (Bai nay e tao file bang 'w', nhung bai khac e tao file tren folder)
with open(WORKDIR/"th13_5.txt", "w") as f:
    f.write(s)

    
print(count("th13_5.txt", char))

```

```python
#W11A6
from pathlib import Path
WORKDIR = Path(".")
def countCharacter(path: str, letter: str):
    count = 0
    file = WORKDIR/path
    with open(file, "r") as f:
        s = f.read()
        s = s.lower()
        for c in s:
            if c == letter:
                count += 1
    return count
#main
letter = "e"
print(countCharacter("th13_6.txt", letter))
```
