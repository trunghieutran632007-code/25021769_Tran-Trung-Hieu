```python
#W11A1
from pathlib import Path
WORKDIR = Path(".")
def Reverse(s):
    s = s[::-1]
    return int(s)
file = WORKDIR/ "th13_1.txt"
with open(file, "r") as f:
    n = f.read().strip()
print(Reverse(n))
```

```python
#W11A2
from pathlib import Path
WORKDIR = Path(".")
def canWin(n):
    if n % 4 == 0:
        return False
    else:
        return True
    
file = WORKDIR/"th13_2.txt"
with open(file, "r") as f:
    n = int(f.read().strip())
print(canWin(n))
```

```python
#W11A3
res = []
from pathlib import Path
WORKDIR = Path(".")
def housesOfHogwarts(path: str):
    file = WORKDIR/path
    with open(file, "r") as f:
        num = int(f.readline().strip())
        for _ in range(num):
            line = f.readline()
            res.append(line)
    return res
#print(housesOfHogwarts("th13_3.txt"))
lstOfName = housesOfHogwarts("th13_3.txt")
for name in lstOfName:
    print(name)


```
