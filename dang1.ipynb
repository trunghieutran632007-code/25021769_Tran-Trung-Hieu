```python
#bai23
def solve_iterative(n):
    """
    Tính F_n bằng phương pháp lặp (hiệu quả hơn cho n lớn).
    Công thức: F_0 = 1, F_1 = 1, F_n = 2*F_(n-1) + F_(n-2)
    """
    if n < 0:
        return "Input n phải là số không âm."
    if n == 0:
        return 1
    if n == 1:
        return 1

    # Khởi tạo hai giá trị đầu
    f0 = 1  # F_0
    f1 = 2  # F_1

    for _ in range(2, n + 1):
        f_n = 2 * f1 - f0
        # Cập nhật giá trị cho lần lặp tiếp theo
        f0 = f1
        f1 = f_n

    return f_n
n = int(input())
print(solve_iterative(n))
```

```python
#bai 54

def prime_factorization(n):
    """
    Phân tích một số nguyên dương n thành các thừa số nguyên tố.
    In ra các thừa số theo thứ tự tăng dần, cách nhau bằng dấu phẩy.
    """
    if n <= 1:
        # Nếu n <= 1, không có thừa số nguyên tố
        return

    factors = []

    # 1. Xử lý thừa số 2
    while n % 2 == 0:
        factors.append(2)
        n //= 2  # Phép chia lấy phần nguyên

    # 2. Xử lý các thừa số nguyên tố lẻ (từ 3 trở đi)
    # Ta chỉ cần kiểm tra đến căn bậc hai của n
    i = 3
    while i * i <= n:
        while n % i == 0:
            factors.append(i)
            n //= i
        i += 2  # Tăng i lên 2 để chỉ kiểm tra số lẻ

    # 3. Xử lý thừa số nguyên tố cuối cùng (nếu n > 1)
    # Nếu sau vòng lặp mà n > 1, n chính là một số nguyên tố lớn
    if n > 1:
        factors.append(n)

    for i in factors:
        print(i,end=' ')


n=int(input())
prime_factorization(n)

```
