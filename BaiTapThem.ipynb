```python
#So Amstrong
def so_Asm(x):
    chuoi = str(x)
    n = len(chuoi)
    temp = x
    sum = 0
    while temp > 0:
        lastnum = temp % 10
        sum += (lastnum) ** n
        temp = temp // 10
    if x == sum:
        return True
    return False

x = int(input())
if so_Asm(x):
    print('so AStr')
else:
    print('kphai')
```

```python
#So chinh phuong
def So_chinh_phuong(x):
    y = x ** 0.5
    y = int(y)
    if y **2 == x:
        return True
    return False

x = int(input())
if So_chinh_phuong(x):
    print('So cp')
else:
    print('kphai')
```

```python
def So_hoan_hao(x):
    # Số hoàn hảo nhỏ nhất là 6, nên x < 6 trả về False ngay (bao gồm số 1)
    if x < 6:
        return False
    
    # Bắt đầu tổng bằng 1 (vì 1 luôn là ước số thực sự)
    tong_uoc = 1 
    
    # Chạy từ 2 đến căn bậc 2 của x
    for i in range(2, int(x ** 0.5) + 1):
        if x % i == 0:
            tong_uoc += i               # Cộng ước i
            # Cộng thêm ước đối xứng (x // i)
            # Thêm điều kiện i != x//i để tránh cộng 2 lần nếu x là số chính phương
            if i != x // i:
                tong_uoc += x // i
                
    return tong_uoc == x

# Phần nhập xuất
try:
    x = int(input("Nhập số: "))
    if So_hoan_hao(x):
        print('Phai')
    else:
        print('kphai')
except ValueError:
    print("Vui lòng nhập số nguyên.")

```
