```python
#W9A1
def Sort(a1, a2):
    new_a = []
    n1 = len(a1)
    n2 = len(a2)
    i = j = 0
    while i < n1 and j < n2:
        if a1[i] < a2[j]:
            new_a.append(a1[i])
            i += 1
        else:
            new_a.append(a2[j])
            j += 1
    while i < n1:
        new_a.append(a1[i])
        i += 1
    while j < n2:
        new_a.append(a2[j])
        j += 1
    return new_a

a1 = list(map(int, input().split()))
a2 = list(map(int, input().split()))
print(Sort(a1, a2))

```

```python
#W9A2
def count_less_equal(matrix, n, m, target):
    count = 0
    row = n - 1
    col = 0
    while row >= 0 and col < m:
        if matrix[row][col] <= target:
            count += (row + 1)
            col += 1
        else:
            row -= 1
    return count  # <--- BẠN THIẾU DÒNG NÀY

def find_kth_num(matrix, k):
    # Thêm kiểm tra nếu ma trận rỗng để tránh lỗi index
    if not matrix or not matrix[0]:
        return None
        
    low = matrix[0][0]
    high = matrix[-1][-1] # Dùng index âm cho gọn và chuẩn Python
    
    while low < high:
        mid = (low + high) // 2
        count = count_less_equal(matrix, len(matrix), len(matrix[0]), mid)
        if count < k:
            low = mid + 1
        else:
            high = mid
    return low

# --- Phần nhập liệu ---
n = int(input("Nhap so hang: "))
m = int(input("Nhap so cot: "))
matrix = []
for i in range(n):
    # Sửa lại câu lệnh input để rõ ràng hơn
    row = list(map(int, input(f"Nhap hang thu {i+1} (cac so cach nhau boi dau cach): ").split()))
    matrix.append(row)

# Bạn nên cho nhập k luôn thay vì fix cứng là 12
k = int(input(f"Nhap so thu k can tim (1 <= k <= {n*m}): "))
print(f"So nho thu {k} la: {find_kth_num(matrix, k)}")
```

```python
#W9A4
def check_abbreviation(a, b):
    m, n = len(a), len(b)
    # dp[i][j] nghĩa là a[:i] có thể biến thành b[:j] không
    dp = [[False] * (n + 1) for _ in range(m + 1)]
    dp[0][0] = True

    # Trường hợp b rỗng, a chỉ đúng nếu a toàn chữ thường (để xóa hết)
    for i in range(1, m + 1):
        if a[i-1].islower():
            dp[i][0] = dp[i-1][0]
        else:
            dp[i][0] = False

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            char_a = a[i-1]
            char_b = b[j-1]

            if char_a == char_b:
                # Ký tự giống nhau (cùng là hoa hoặc cùng logic khớp)
                dp[i][j] = dp[i-1][j-1]
            elif char_a.upper() == char_b:
                # a là thường, viết hoa lên thì khớp b -> Có thể khớp HOẶC xóa a
                dp[i][j] = dp[i-1][j-1] or dp[i-1][j]
            elif char_a.islower():
                # a là thường nhưng không khớp -> Chỉ có thể xóa a
                dp[i][j] = dp[i-1][j]
            else:
                # a là hoa và không khớp -> False
                dp[i][j] = False
                
    return "YES" if dp[m][n] else "NO"

# Phần Main xử lý Q truy vấn
try:
    q = int(input()) # Đọc số lượng câu hỏi
    for _ in range(q):
        a = input()
        b = input()
        print(check_abbreviation(a, b))
except:
    pass
```

```python
#W9A5
def check_list(lst1, lst2):
    len1 = len(lst1)
    len2 = len(lst2)
    
    for i in range(len2 - len1 + 1):
        if lst2[i : i + len1] == lst1:
            return "YES"
            
    return "NO"
a1 = list(map(int, input().split()))
a2 = list(map(int, input().split()))
print(check_list(a1,a2))


```

```python
#W9A6
def next_permutation(arr):
    n = len(arr)
    i = n - 2
    while i >= 0 and arr[i] >= arr[i + 1]:
        i -= 1
    if i == -1:
        return arr[::-1]
    j = n - 1
    while arr[j] <= arr[i]:
        j -= 1
        
    arr[i], arr[j] = arr[j], arr[i]
    
    left = i + 1
    right = n - 1
    while left < right:
        arr[left], arr[right] = arr[right], arr[left]
        left += 1
        right -= 1
        
    return arr

n = int(input())
a = list(map(int, input().split()))
print(*next_permutation(a))
```

```python
#W9A7
def canPlaceFlowers(flowerbed, k):
    count = 0
    length = len(flowerbed)
    for i in range(length):
        if flowerbed[i] == 0:
            left_empty = (i == 0) or (flowerbed[i - 1] == 0)
            right_empty = (i == length - 1) or (flowerbed[i + 1] == 0)
            
            if left_empty and right_empty:
                flowerbed[i] = 1
                count += 1
                if count >= k:
                    return True
    return count >= k


    
```

```python
#W9A8
def Di_chuyen(s):
    Toa_do = [0,0]
    path = []
    for c in s:
        if c == "R":
            Toa_do[0] += 1
            path.append(Toa_do[:])
        if c == "L":
            Toa_do[0] -= 1
            path.append(Toa_do[:])
        if c == "U":
            Toa_do[1] += 1
            path.append(Toa_do[:])
        if c == "D":
            Toa_do[1] -= 1
            path.append(Toa_do[:])
    return path
def Di_qua(path):
    visited = {(0,0)}
    path_tup = [tuple(x) for x in path]
    for point in path_tup:
        if point in visited:
            return True
        else:
            visited.add(point)
    return False

s = "UURRDL"
print(Di_chuyen(s))
print(Di_qua(Di_chuyen(s)))

```

```python
#W9A9
def Median(nums):
    # --- LƯỢT 1: Sắp xếp các cặp đôi ---
    # So sánh 1: Sắp xếp cặp (nums[0], nums[1]) sao cho nums[0] < nums[1]
    if nums[0] > nums[1]:
        nums[0], nums[1] = nums[1], nums[0]
        
    # So sánh 2: Sắp xếp cặp (nums[2], nums[3]) sao cho nums[2] < nums[3]
    if nums[2] > nums[3]:
        nums[2], nums[3] = nums[3], nums[2]
        
    # --- LƯỢT 2: Loại bỏ phần tử nhỏ nhất (Min) ---
    # So sánh 3: So sánh hai số nhỏ của hai cặp trên (nums[0] và nums[2])
    # Mục tiêu: Đưa số nhỏ hơn về nums[0]. Lúc này nums[0] sẽ nhỏ hơn nums[1], nums[2], nums[3].
    # Nó chắc chắn là số nhỏ nhất hoặc nhì, không thể là trung vị (số lớn thứ 3). Ta loại nums[0].
    if nums[0] > nums[2]:
        nums[0], nums[2] = nums[2], nums[0]
        # Quan trọng: Phải tráo luôn cả "partner" của nó để giữ quan hệ lớn-nhỏ đã thiết lập
        nums[1], nums[3] = nums[3], nums[1]

    # --- LƯỢT 3: Xử lý số lẻ (nums[4]) và tìm số nhỏ thứ nhì ---
    # Lúc này nums[4] chưa được đụng đến. Ta ghép nó với nums[1] (số còn lại của cặp chứa min bị loại).
    # So sánh 4: Sắp xếp cặp (nums[1], nums[4])
    if nums[1] > nums[4]:
        nums[1], nums[4] = nums[4], nums[1]
    
    # Lúc này ta có 2 cặp đã sắp xếp thứ tự nội bộ: (nums[1] < nums[4]) và (nums[2] < nums[3])
    # Ta cần tìm số nhỏ nhất trong 4 số còn lại (để loại đi, vì đó là số nhỏ thứ 2 tổng thể).
    # So sánh 5: So sánh hai đầu nhỏ của hai cặp (nums[1] và nums[2])
    if nums[1] > nums[2]:
        nums[1], nums[2] = nums[2], nums[1]
        nums[4], nums[3] = nums[3], nums[4] # Tráo partner tương ứng

    # Sau bước 5, nums[1] là số nhỏ nhất trong các số còn lại.
    # Tổng kết: nums[0] là Min (hạng 1), nums[1] là hạng 2.
    # Trung vị (hạng 3) sẽ là số nhỏ hơn trong hai ứng cử viên còn lại ở vị trí thấp: nums[2] và nums[4].
    # (Lưu ý: nums[3] lớn hơn nums[2] nên nums[3] ít nhất là hạng 4, không thể là hạng 3 nếu nums[2] < nums[4]).

    # --- LƯỢT 4: Chốt hạ ---
    # So sánh 6: Tìm min giữa nums[2] và nums[4]
    if nums[2] < nums[4]:
        return nums[2]
    else:
        return nums[4]

# --- Kiểm tra thử ---
# Test case ngẫu nhiên
if __name__ == "__main__":
    danh_sach = [12, 4, 56, 1, 9] 
    print(f"Danh sách: {danh_sach}")
    print(f"Trung vị tìm được: {Median(danh_sach)}")
    # Output mong đợi: 9 (vì thứ tự là 1, 4, 9, 12, 56)
```
