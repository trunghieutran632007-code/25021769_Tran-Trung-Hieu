```python
#W6A5
a = list(map(int, input().split()))
neg = []
pos = []
zeros = []
for i in range(len(a)):
    if a[i] > 0:
        pos.append(a[i])
    elif a[i] < 0:
        neg.append(a[i])
    else:
        zeros.append(a[i])
numbers = {
    'Pos': pos,
    'Neg': neg,
    'zeros': zeros,
}
print(numbers)
```
