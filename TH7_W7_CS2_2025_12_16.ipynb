```python
#W7A1
def Binary_search(arr, target):
    left = 0
    right = len(arr)-1

    while left <= right:
        mid = (left + right) // 2
        mid_val = arr[mid]
        if mid_val == target:
            return mid #tra ve vi tri da tim duoc
        elif mid_val < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1



```
