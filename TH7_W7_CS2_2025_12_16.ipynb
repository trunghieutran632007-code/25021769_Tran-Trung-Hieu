```python
#W7A1
def Binary_search(arr, target):
    left = 0
    right = len(arr)-1

    while left <= right:
        mid = (left + right) // 2
        mid_val = arr[mid]
        if mid_val == target:
            return mid #tra ve vi tri da tim duoc
        elif mid_val < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1


arr = [10,15,1,2,4]
arrnew = arr.sort()
Binary_search(arr, 15)
print(arr)
print(arrnew)

```

```python
#W7A3
def Bubble_sort(arr):
    for i in range(len(arr)):
        swap = False
        for j in range(len(arr)-i-1):
            if arr[j]>arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swap = True
        if not swap:
            break
    return arr

def Selection_sort(arr):
    for i in range(len(arr)):
        minInd = i
        for j in range(i+1, len(arr)):
            if arr[minInd]>arr[j]:
                minInd = j
        if minInd != i:
            arr[minInd], arr[i] = arr[i], arr[minInd]

    return arr


arr = list(map(int, input().split()))
print(Bubble_sort(arr))
print(Selection_sort(arr))            
```

```python
#W7A4
def max_Occur(arr):
    myDict = {}
    for num in arr:
        if num in myDict:
            myDict[num] += 1
        else:
            myDict[num] = 1
    max = 0
    for key, value in myDict.items():
        if value > max:
            max = value
            maxOccur = key
    
    for num in arr:
        if num == maxOccur:
            return f'{num} xuat hien nhieu nhat,som nhat, {max} lan'

arr = list(map(int, input().split()))
print(max_Occur(arr))
```

```python
#W7A5
def Thoa_man_gt(arr, x):
    count = 0
    for i in range(len(arr)):
        for j in range(i+1, len(arr)):
            if (arr[i]+arr[j]) == x:
                count += 1
    return count

arr = list(map(int, input().split()))
x = int(input())
print(Thoa_man_gt(arr, x))
```

```python
#W7A17
#Bài tập về thuật toán Boyer-Moore Voting tìm phần tử chiếm hơn nửa trong mảng
def Tim_phan_tu_da_so(arr):
    count = 0
    candidate = None

    for num in arr:
        if count == 0:
            candidate = num
            count = 1
        elif num == candidate:
            count += 1
        else:
            count -= 1

    real_count = 0
    for num in arr:
        if num == candidate:
            real_count += 1
    if real_count > len(arr)//2:
        return candidate
    else:
        return ' '
    
arr = list(map(int, input().split()))
print(Tim_phan_tu_da_so(arr))
```

```python
#W7A21 #Kết hợp chương trình ngày tháng hợp lệ
def Nam_nhuan(nam):
    return (nam%4 == 0 and nam%100 != 0) or (nam%400 == 0)
def Ngay_thang_hop_le(ngay, thang, nam):
    m = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if Nam_nhuan(nam):
        m[2] = 29
    if ngay < 1 or nam < 1 or thang>12 or thang < 1 or ngay > m[thang]:
        return False
    else:
        return True
def Ngay_thu_bao_nhieu(ngay, thang, nam):
    m = [0, 31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
    if Nam_nhuan(nam):
        m[2] = 29
    ngay_thu = 0
    if thang >= 2:
        for i in range(1, thang):
            ngay_thu += m[i]
    return ngay_thu+ngay


while True:
    try:
        ngay = int(input())
        thang = int(input())
        nam = int(input())
        if Ngay_thang_hop_le(ngay, thang, nam):
            print('Valid')
            print(f'Ngay thu {Ngay_thu_bao_nhieu(ngay, thang, nam)}')
            break
    except ValueError:
        print('Nhap ngay thang hop le')





```

```python
#W7A23
def Dem_ten(s):
    count = 0
    new_s = ""
    #Bo di dau cau
    for c in s:
        if c.isalpha() or c == " ":
            new_s += c
        else:
            new_s += " "
    new_s = new_s.split()
    for word in new_s:
        if len(word) > 2:
            dk_Hoa = word[0].isupper()
            dk_viet_thuong = word[1:]

            if dk_Hoa and dk_viet_thuong:
                count += 1
    return count

s =  input()
print(Dem_ten(s))
```

```python
#W7A24
def Xoa_tu_giong_nhau(s):
    lst_of_word = s.split()
    index = []
    for i in range(0, len(lst_of_word)-1):
        if lst_of_word[i] in lst_of_word[i+1]:
            index.append(i)

    for j in index:
        lst_of_word.pop(j)
    new_s = " ".join(lst_of_word)
    return new_s

s = input()
print(Xoa_tu_giong_nhau(s))
            
```

```python
#W7A25
def largest_num(num_s, k):
    stack = []
    remove = len(num_s)-k
    for digit in num_s:
        while remove > 0 and stack and stack[-1] < digit:
            stack.pop()
            remove -= 1
        stack.append(digit)
    res = ''.join(stack[:k])
    return res

s = input()
k = int(input())
print(largest_num(s, k))
```

```python
#W7A26
def Tach_so(s):
    num_s = ""
    lst_of_nums = []
    for i in range(len(s)-1):
        if s[i].isdigit():
            num_s += s[i]
            if not s[i+1].isdigit():

                lst_of_nums.append(num_s)
                num_s = ""
            else:
                if i + 1 == len(s)-1:
                    num_s += s[i+1]
                    lst_of_nums.append(num_s)
    lst_of_nums = [int(x) for x in lst_of_nums]
    return f'{sum(lst_of_nums)} {sum(lst_of_nums)/len(lst_of_nums)}'
s = input()
print(Tach_so(s))

```

```python
#W7A26 cách 2 (by Gemini)
def Tach_so_manual(s):
    lst_of_nums = []
    temp_num = ""
    
    # Thêm một ký tự không phải số vào cuối chuỗi để đảm bảo số cuối cùng luôn được xử lý
    # Đây là kỹ thuật "lính canh" (sentinel)
    s += " " 
    
    for char in s:
        if char.isdigit():
            temp_num += char # Nếu là số, gom vào biến tạm
        else:
            if temp_num != "": # Nếu gặp chữ và biến tạm đang có số
                lst_of_nums.append(int(temp_num)) # Lưu số đó lại
                temp_num = "" # Reset biến tạm
                
    if not lst_of_nums:
        return "0 0"
        
    return f'{sum(lst_of_nums)} {sum(lst_of_nums)/len(lst_of_nums)}'
s = input()
print(Tach_so_manual(s))
```
